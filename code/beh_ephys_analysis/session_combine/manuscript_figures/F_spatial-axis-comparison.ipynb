# F_spatial-axis-comparison.ipynb

## Purpose
Integrates three datasets to compare cellular organization axes: electrophysiology waveform features, MERFISH spatial transcriptomics (~2,300 cells, 315 genes), and retrograde tracing from 37 brains. Fits a primary spatial axis to each dataset using CCA (waveform, gene expression) or LDA (projection targets) and compares principal spatial gradients across data modalities. Projects LC neuron coordinates onto all three axes to test whether functional behavioral tuning is spatially organized.

## Prerequisites

Required input files:
- `{manuscript_fig_prep_dir}/combined_unit_tbl/combined_unit_tbl.pkl` — combined unit table with CCF coordinates and session/unit IDs
- `{manuscript_fig_prep_dir}/waveforms_np/combined_features.csv` — waveform PC features per unit
- `{data_dir}/LC_percentile_meshes/core_mesh.obj` — LC core mesh for anatomical overlays
- `{data_dir}/merfish_data/adata_mer_subset_2_2k.h5ad` — MERFISH spatial transcriptomics (~2,300 cells, 315 genes)
- `{data_dir}/LC_retro/manual_proofread_ccf_37brains.csv` — manually proofread retrograde tracing CCF coordinates (37 brains)
- `{manuscript_fig_dir}/F_ephys_beh_action&outcome/features_combined_beh_all.pkl` — combined behavioral feature table

## Figure Panels

- **Figure5d**: Respond-vs-ignore encoding along axonal-projection and gene-expression axes (scatter, regression)
- **Figure5h**: Switch-vs-stay encoding along axonal-projection and gene-expression axes (scatter, regression)

## Configuration and Mesh Setup

This section initializes output directories and loads the LC mesh geometry. The mesh vertices are transformed to the LPS (Left-Posterior-Superior) coordinate system relative to bregma, which establishes the anatomical reference frame for all spatial analyses. Mesh contours are pre-computed for three anatomical planes (sagittal, horizontal, coronal) to overlay on spatial plots.

```python
target_folder = os.path.join(capsule_dirs["manuscript_fig_dir"], 'F_spatial_axis_compare')
os.makedirs(target_folder, exist_ok=True)
```

```python
ml, ap, dv = 0, 1, 2
bregma_LPS_mm = np.array([-5.7, 5.4, -0.45])  # in mm
planes = {'sag': [ap, dv], 'hor': [ml, ap], 'cor': [ml, dv]}
mesh_file = str(capsule_directories()['data_dir']) + '/LC_percentile_meshes/core_mesh.obj' 
mesh = load_mesh(mesh_file)
mesh_vertices = np.array(mesh.vertices)
mesh_vertices = mesh_vertices/1000
if 'scratch' in mesh_file:
    mesh_vertices = (mesh_vertices - np.array([216, 18, 228]))*25/1000
    mesh_vertices_lps = pir_to_lps(mesh_vertices) # convert to lps
    mesh_vertices_mm = mesh_vertices_lps
else:
    mesh_vertices_mm = mesh_vertices.copy()
    mesh_vertices_mm[:, 2] = -mesh_vertices_mm[:, 2]
    mesh_vertices_mm[:, 0] = -mesh_vertices_mm[:, 0]
    mesh_vertices_mm = mesh_vertices_mm - bregma_LPS_mm
mesh_vertices_mm[:, ml] = np.abs(mesh_vertices_mm[:, ml])

mesh_contours = {
    plane_name: project_to_plane(mesh_vertices_mm, plane_axes, pitch=0.02, margin=0.5)
    for plane_name, plane_axes in planes.items()
}
```

## Dataset X: Electrophysiology Waveform Features

Load the combined electrophysiology unit table and waveform features from preprocessed files. Waveform features quantify the shape properties of extracellular action potentials (e.g., peak-to-trough ratios, slopes, symmetry). These features are merged with CCF coordinates for each recorded unit. The primary analysis focuses on the first principal component of waveform features (wf_pc_1), which captures the dominant axis of waveform variation across the LC population.

```python
# load waveform info and location
with open(os.path.join(capsule_dirs["manuscript_fig_prep_dir"], 'combined_unit_tbl', 'combined_unit_tbl.pkl'), 'rb') as f:
    combined_tagged_units = pickle.load(f)
waveform_file = os.path.join(capsule_dirs["manuscript_fig_prep_dir"], 'waveforms_np', 'combined_features.csv')
wf_combined_features = pd.read_csv(waveform_file)

# merge at sessoin and unit id
wf_combined_features = wf_combined_features.merge(combined_tagged_units[['session', 'unit', 'x_ccf', 'y_ccf', 'z_ccf']], on=['session', 'unit'], how='left')
wf_feature_focus = 'wf_pc_1'
wf_features_focus = [
                'post_w', 
                'trough_post_ratio_1D', 
                'post_trough_slope', 
                'pre_slope',
                'symmetry_slope_div_log',
                'symmetry_trough_dis', 
                'symmetry_inte_div_log', 
                ]

ccf_wf = wf_combined_features[['x_ccf', 'y_ccf', 'z_ccf']].values
ccf_wf = ccf_wf - bregma_LPS_mm
ccf_wf[:, ml] = np.abs(ccf_wf[:, ml])  # mirror ML to left

filter_wf = ~np.isnan(wf_combined_features[wf_feature_focus].values) & ~np.any(np.isnan(ccf_wf), axis=1)
feature_x = wf_combined_features[wf_feature_focus].values[filter_wf]
features_x = wf_combined_features[wf_features_focus].values[filter_wf]
coord_x = ccf_wf[filter_wf].copy()
```

## Dataset Y: MERFISH Spatial Transcriptomics

Load MERFISH spatial transcriptomics data (~2,200 cells, 315 genes). The data undergoes library-size normalization and filtering to remove ventral cells (DV > 220 µm). Spatial coordinates are converted to the same LPS reference frame as the electrophysiology data. The gene expression matrix is reduced to its first principal component via PCA, capturing the dominant transcriptional gradient across LC cells.

```python
# load gene expression info
MERFISH_DATA_DIR  = str(capsule_directories()['data_dir']) + '/merfish_data'

def load_and_preprocess_data():
    """Load  MERFISH data, perform initial preprocessing""" # not laoding mesh for this
    
    
    print("Loading MERFISH data...")
    filename = os.path.join(MERFISH_DATA_DIR, 'adata_mer_subset_2_2k.h5ad')    
    adata_mer = sc.read_h5ad(filename)
    librarysize = np.sum(adata_mer.X, 1)
    normed_x = 1000*adata_mer.X/librarysize[:,None]
    adata_mer.X = normed_x

    # some removal of the ventral cell
    y = adata_mer.obsm["spatial"][:, 1]
    keep = y <= 220
    print(f"Removing {(~keep).sum()} cells with DV axis > 220")  # this shoudl show 41
    adata_mer = adata_mer[keep].copy()
    print(f"After filtering: {adata_mer.shape}")  # this should be 2221
    return _, _, adata_mer

_, _, adata_mer = load_and_preprocess_data()

features_y = adata_mer.X
coord_y = adata_mer.obsm['spatial']
coord_y_mm = ccf_pts_convert_to_mm(coord_y)
coord_y_mm= pir_to_lps(coord_y_mm)
coord_y_mm[:, ml] = np.abs(coord_y_mm[:, ml])  # mirror ML to left
coord_y_mm[:, ap] = -coord_y_mm[:, ap]  # flip AP to match atlas orientation
filter = (coord_y_mm[:, dv]>-10)
features_y = features_y[filter]
coord_y = coord_y_mm[filter, :]

# extract PC1 from features_y as feature y
pca = PCA(n_components=1)
feature_y_pc1 = pca.fit_transform(features_y).flatten()
feature_y = feature_y_pc1
```

## Dataset Z: Retrograde Tracing from Projection Targets

Load manually proofread retrograde tracing data from 18 brains. Labeled LC neurons are categorized by their injection target region (e.g., cortex, striatum, thalamus). CCF coordinates are transformed to match the LPS reference frame. Thalamus-projecting neurons and outliers (DV < -5 mm) are excluded. Categorical injection regions are converted to numeric codes for LDA analysis.

```python
# load retro data
data_dir = str(capsule_directories()['data_dir']) + '/LC_retro/manual_proofread_ccf_37brains.csv'
retro_ccf = pd.read_csv(data_dir)
bregma_pixel_retro = np.array([228, 18, 216])
ccf_retro = (retro_ccf[['x', 'y', 'z']].values - bregma_pixel_retro) * 25 / 1000  # convert to mm
ccf_retro[:, [dv, ap]] = ccf_retro[:, [ap, dv]]  # Mirror ML to left
ccf_retro[:, dv] = -ccf_retro[:, dv]  # Mirror ML to left
ccf_retro[:, ml] = np.abs(ccf_retro[:, ml])  # Mirror ML to left
filter = (retro_ccf['injection_region'].values != 'TH') & (ccf_retro[:, dv]>-5)
feature_z = retro_ccf['injection_region'].values[filter]
cat = pd.Categorical(feature_z)
feature_z = cat.codes
categories = cat.categories
coord_z = ccf_retro[filter, :]
```

## Fit Spatial Axes with Bootstrap Confidence Intervals

Fit spatial axes for all three datasets using appropriate methods:

- **Dataset X (waveform)**: CCA between 7 waveform features and 3D coordinates (2000 bootstrap iterations)
- **Dataset Y (gene expression)**: CCA between 315 gene expression values and 3D coordinates
- **Dataset Z (projection targets)**: LDA to find the direction that best separates injection target categories

For each dataset, compute bootstrap distributions to quantify directional uncertainty. All bootstrap vectors are hemisphere-aligned to the observed axis for consistent orientation. The spatial bin resampling option is available but currently disabled (bootstrap_within_spatial_bins=False).

```python
n_boot = 2000
res_boot_x = bootstrap_spatial_axis_cca(
    features_x,
    coord_x,
    n_boot=n_boot,
    align_to_observed=True,
    seed=16,
    bootstrap_within_spatial_bins=False,
    spatial_bin_counts=(2, 3, 3),

)
axis_x = res_boot_x["axis_unit"]
axis_boot_x = res_boot_x["axis_boot"]
res_boot_y = bootstrap_spatial_axis_cca(
    features_y,
    coord_y,
    n_boot=n_boot,
    align_to_observed=True,
    seed=2,
    bootstrap_within_spatial_bins=False,
    spatial_bin_counts=(2, 3, 3),
)
axis_y = res_boot_y["axis_unit"]
axis_boot_y = res_boot_y["axis_boot"]

res_boot_z = bootstrap_spatial_axis_LDA(
    feature_z,
    coord_z,
    n_boot=n_boot,
    seed=0,
    align_to_observed=True,
    return_full=True,
)
if np.dot(res_boot_z["axis_unit"], axis_x) < 0:
    res_boot_z["axis_unit"] = -res_boot_z["axis_unit"]
    res_boot_z["beta"] = -res_boot_z["beta"]
    if "beta_boot" in res_boot_z:
        res_boot_z["beta_boot"] = -res_boot_z["beta_boot"]
    if "axis_boot" in res_boot_z:
        res_boot_z["axis_boot"] = -res_boot_z["axis_boot"]

axis_z = res_boot_z["axis_unit"]
axis_boot_z = res_boot_z["axis_boot"]
```

**Pairwise statistical comparisons**: Test whether the three spatial axes differ significantly using tangent-plane projections and bootstrap-based inference. For each pair (X-Y, X-Z, Y-Z), compute the 2D deviation on the tangent plane perpendicular to one reference axis, then assess whether this deviation is statistically distinguishable from zero using Wald statistics and bootstrap p-values.

## Correlate Behavioral Features with Spatial Axes

Project each LC neuron's 3D coordinates onto all three spatial axes (waveform, gene expression, projection target). For six behavioral features (baseline/response tuning for hit-miss, outcome, switch), compute Pearson correlations between spatial projections and behavioral tuning strengths.

Generate a comprehensive figure matrix (6 features × 3 axes = 18 panels) showing:
- Scatter plots of behavioral feature vs. spatial projection
- Linear regression fits with 95% confidence intervals
- Pearson r and p-values (highlighted in red if p < 0.05)

This analysis tests whether functional properties of LC neurons are organized along any of the anatomical gradient axes identified from waveform, gene expression, or connectivity data.
